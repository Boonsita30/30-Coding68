```python
print("💗🌸" + "═"*40 + "🌸💗")
print("💖      โปรแกรมคำนวณค่าไฟฟ้า      💖")
print("💗🌸" + "═"*40 + "🌸💗")

name = input("🌷 ยินดีต้อนรับเข้าสู่โปรแกรมคำนวณค่าไฟฟ้าค่ะ\n💞 ผู้ใช้ชื่ออะไรคะ: ")

print("\n🌸 กรุณากรอกจำนวนหน่วยค่าไฟ 🌸")
unit1 = int(input("💗 ใส่เลข 1-25: "))
unit2 = int(input("💗 ใส่เลข 1-30: "))

total = unit1 + unit2

print("\n💖🌷" + "─"*40 + "🌷💖")
print(f"🌸 คุณ {name} ใช้ไฟฟ้ารวมทั้งหมด: {total} หน่วย 💕")

if total <= 50:
    print("💗 ค่าไฟฟ้าของคุณ หน่วยละ 2.50 บาท 🌷")
else:
    print("💗 ค่าไฟฟ้าของคุณ หน่วยละ 5.00 บาท 🌷")

print("💖🌸" + "═"*40 + "🌸💖")
print("💞 ขอบคุณที่ใช้บริการค่ะ 💞")
print("💗🌸" + "═"*40 + "🌸💗")
```
